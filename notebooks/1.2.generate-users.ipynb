```python
import csv
import hashlib
import secrets
import string
from pathlib import Path

N_USERS = 50
PASSWORD_LENGTH = 8
OUT_HASHED = Path("data/users_hashed.csv")
OUT_PLAIN = Path("data/users_plain.csv")

# caractere-set alfanumérico (maiusculas, minusculas e dígitos)
alphabet = string.ascii_letters + string.digits

def gen_password(length=8):
    return ''.join(secrets.choice(alphabet) for _ in range(length))

def hash_password(pw: str) -> str:
    return hashlib.sha256(pw.encode("utf-8")).hexdigest()

def main():
    users = []
    for i in range(1, N_USERS + 1):
        username = f"user{i:02d}"   # user01 ... user50
        pw = gen_password(PASSWORD_LENGTH)
        pw_hash = hash_password(pw)
        users.append((username, pw, pw_hash))

    # gravar arquivo com hashes (usado pela app)
    with OUT_HASHED.open("w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["username", "password_hash"])
        for u, _, h in users:
            writer.writerow([u, h])

    # gravar arquivo com senhas em texto puro (apenas para admin)
    with OUT_PLAIN.open("w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["username", "password_plain"])
        for u, pw, _ in users:
            writer.writerow([u, pw])

    print(f"Gerados {N_USERS} usuários.")
    print(f"Arquivo com hashes: {OUT_HASHED.resolve()}")
    print(f"Arquivo com senhas em texto puro (guarde com segurança): {OUT_PLAIN.resolve()}")
```

```python
main()
```
