```python
# normalize_assignments.py
import pandas as pd

INPUT = "data/assignments.csv"
OUTPUT = "data/assignments_normalized.csv"

# quantos dígitos usar (ex.: 2 -> user01, user02 ... até user99)
PAD_WIDTH = 2

df = pd.read_csv(INPUT, dtype=str)

if "annotator_id" not in df.columns or "frase_id" not in df.columns:
    raise ValueError("assignments.csv precisa ter colunas 'annotator_id' e 'frase_id'.")

# garante número inteiro para cada annotator_id
df["annotator_num"] = df["annotator_id"].astype(str).str.extract(r"(\d+)")[0].astype(int)

# cria a coluna username padronizada
df["username"] = df["annotator_num"].apply(lambda x: f"user{str(x).zfill(PAD_WIDTH)}")

# salva só o que interessa
df_out = df[["username", "frase_id"]].rename(columns={"username": "annotator_id"})
df_out.to_csv(OUTPUT, index=False)

print(f"Arquivo normalizado salvo em: {OUTPUT}")
```
