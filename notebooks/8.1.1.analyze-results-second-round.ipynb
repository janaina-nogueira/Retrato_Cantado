```python
# import ast

# awa = assignments_with_annotations.copy()
# awa['annotation_list'] = awa['annotation'].apply(lambda x: ast.literal_eval(x) if x != 'not_annotated' else [])

# annotation_statistics_df = awa.groupby('frase_id')['annotation_list'].apply(list).reset_index()
# annotation_statistics_df['total_annotations'] = annotation_statistics_df['annotation_list'].apply(lambda x: sum(len(ann) for ann in x))
# def count_annotation(ann_list, ann_type):
#     return sum(ann.count(ann_type) for ann in ann_list)

# for ann_type in ['identidade', 'aparencia', 'papel_social', 'habilidades', 'valores', 'comportamentos', 'outros']:
#     annotation_statistics_df[ann_type] = annotation_statistics_df['annotation_list'].apply(lambda x: count_annotation(x, ann_type))
# annotation_statistics_df['valid_annotations'] = annotation_statistics_df['total_annotations']
# annotation_statistics_df = annotation_statistics_df.drop(columns=['annotation_list'])
# annotation_statistics_df


import ast
from collections import Counter

awa = assignments_with_annotations.copy()

# Convert annotation strings to lists, turning 'not_annotated' into empty lists
awa['annotation_list'] = awa['annotation'].apply(
    lambda x: ast.literal_eval(x) if x != 'not_annotated' else []
)

# Group by frase_id and gather all annotation lists per frase
annotation_statistics_df = (
    awa.groupby('frase_id')['annotation_list']
       .apply(list)
       .reset_index()
)

# Total annotations = total number of items across all lists
annotation_statistics_df['total_annotations'] = annotation_statistics_df['annotation_list'].apply(
    lambda lists: sum(len(lst) for lst in lists)
)

# Count specific annotations
def count_annotation(list_of_lists, ann_type):
    return sum(lst.count(ann_type) for lst in list_of_lists)

annotation_types = [
    'identidade', 'aparencia', 'papel_social', 'habilidades',
    'valores', 'comportamentos', 'outros'
]

for ann_type in annotation_types:
    annotation_statistics_df[ann_type] = annotation_statistics_df['annotation_list'].apply(
        lambda lists: count_annotation(lists, ann_type)
    )

# valid annotations = same as total_annotations (since 'not_annotated' → [])
annotation_statistics_df['valid_annotations'] = annotation_statistics_df['total_annotations']

# Majority vote calculation
def get_majority_vote(list_of_lists):
    flat = [item for sublist in list_of_lists for item in sublist]
    if not flat:
        return None
    counts = Counter(flat)
    return counts.most_common(1)[0][0]

annotation_statistics_df['majority_vote'] = annotation_statistics_df['annotation_list'].apply(
    get_majority_vote
)

# Agreement rate = % of annotators who assigned the majority vote
def agreement_rate(list_of_lists, majority_vote):
    if majority_vote is None:
        return 0.0
    total_annotators = len(list_of_lists)
    if total_annotators == 0:
        return 0.0
    agreeing = sum(1 for lst in list_of_lists if majority_vote in lst)
    return agreeing * 100 / total_annotators

annotation_statistics_df['agreement_rate'] = annotation_statistics_df.apply(
    lambda row: agreement_rate(row['annotation_list'], row['majority_vote']),
    axis=1
)

# Remove raw annotation list
annotation_statistics_df = annotation_statistics_df.drop(columns=['annotation_list'])

annotation_statistics_df
```

```python
import plotly.express as px

annotated_sentences = annotation_statistics_df_full[annotation_statistics_df_full['valid_annotations'] > 0]
total_sentences = annotation_statistics_df_full['frase_id'].nunique()
summary_counts = pd.DataFrame({
    'total_frases': [total_sentences],
    'frases_anotadas': [len(annotated_sentences)],
    'taxa_cobertura_pct': [
        (len(annotated_sentences) / total_sentences * 100) if total_sentences else float('nan')
    ]
})
summary_counts.round(2)
```

```python
category_counts = (
    annotated_sentences['majority_vote']
    .dropna()
    .value_counts()
    .rename_axis('categoria')
    .reset_index(name='frases')
    .sort_values('frases', ascending=False)
)
category_counts
```

```python
# if not category_counts.empty:
#     fig = px.bar(
#         category_counts,
#         x='categoria',
#         y='frases',
#         text='frases',
#         title='Distribuição de frases anotadas por categoria',
#     )
#     fig.update_traces(texttemplate='%{text}', textposition='outside')
#     fig.update_layout(
#         yaxis_title='Número de frases',
#         xaxis_title='Categoria',
#         uniformtext_minsize=12,
#         uniformtext_mode='hide'
#     )
#     fig.show()
# else:
#     print('Sem categorias anotadas para exibir.')

LABELS = {
    "identidade": "Identidade",
    "aparencia": "Aparência",
    "papel_social": "Papel social",
    "habilidades": "Habilidades/Competências",
    "valores": "Valores/Crenças",
    "comportamentos": "Comportamentos",
    "outros": "Outros",
}

if not category_counts.empty:

    # Ordena pelas categorias mais frequentes
    category_counts = category_counts.sort_values('frases', ascending=False).copy()

    # Mapeia nomes amigáveis
    category_counts['label_categoria'] = category_counts['categoria'].map(LABELS)

    fig = px.bar(
        category_counts,
        x='categoria',
        y='frases',
        text='frases',   # <-- agora o texto é só o número
        title='Distribuição de frases anotadas por categoria',
        color='categoria',
        color_discrete_sequence=px.colors.qualitative.Set2
    )

    fig.update_traces(
        texttemplate='%{text}',
        textposition='outside',
        hovertemplate='<b>%{label_categoria}</b><br>Frases: %{y}<extra></extra>'
    )

    fig.update_layout(
        template='simple_white',
        showlegend=False,  # <-- remove legenda
        yaxis_title='Número de frases',
        xaxis_title='Categoria',
        uniformtext_minsize=12,
        uniformtext_mode='hide',
        title_x=0.5,
        margin=dict(l=40, r=20, t=60, b=80),
        height=500,
        xaxis=dict(
            tickangle=-30,
            tickmode='array',
            tickvals=category_counts['categoria'],
            ticktext=category_counts['label_categoria'],  # nomes amigáveis no eixo X
            categoryorder='total descending'
        )
    )

    # Mais espaço para o número no topo
    max_y = category_counts['frases'].max()
    fig.update_yaxes(range=[0, max_y * 1.25])

    fig.show()

else:
    print('Sem categorias anotadas para exibir.')
```

```python
agreement_series = annotated_sentences['agreement_rate'].dropna()
agreement_summary = pd.DataFrame({
    'agreement_medio_pct': [agreement_series.mean()],
    'agreement_desvio_pct': [agreement_series.std()],
    'frases_consideradas': [len(agreement_series)]
})
agreement_summary.round(2)
```

```python
agreement_by_class = (
    annotated_sentences[annotated_sentences['majority_vote'].notna()]
    .groupby('majority_vote')['agreement_rate']
    .agg(['count', 'mean', 'std'])
    .reset_index()
    .rename(columns={
        'majority_vote': 'categoria',
        'count': 'frases',
        'mean': 'agreement_medio_pct',
        'std': 'agreement_desvio_pct'
    })
    .sort_values('agreement_medio_pct', ascending=False)
)
agreement_by_class.round({
    'agreement_medio_pct': 2,
    'agreement_desvio_pct': 2
})
```

```python
# if not agreement_by_class.empty:
#     fig = px.bar(
#         agreement_by_class,
#         x='categoria',
#         y='agreement_medio_pct',
#         error_y='agreement_desvio_pct',
#         text='agreement_medio_pct',
#         title='Agreement médio por categoria (com desvio padrão)',
#     )
#     fig.update_traces(texttemplate='%{text:.2f}%', textposition='outside')
#     y_max = agreement_by_class['agreement_medio_pct'].max()
#     y_err = agreement_by_class['agreement_desvio_pct'].fillna(0).max()
#     fig.update_layout(
#         yaxis_title='Agreement médio (%)',
#         xaxis_title='Categoria',
#         yaxis_range=[0, min(100, y_max + y_err * 1.1)],
#         uniformtext_minsize=10,
#         uniformtext_mode='hide'
#     )
#     fig.show()
# else:
#     print('Sem dados de agreement por categoria para exibir.')

LABELS = {
    "identidade": "Identidade",
    "aparencia": "Aparência",
    "papel_social": "Papel social",
    "habilidades": "Habilidades/Competências",
    "valores": "Valores/Crenças",
    "comportamentos": "Comportamentos",
    "outros": "Outros",
}

if not agreement_by_class.empty:
    # Ordena pelas categorias com maior agreement médio
    agreement_by_class = agreement_by_class.sort_values(
        'agreement_medio_pct', ascending=False
    ).copy()

    # Nome amigável da categoria
    agreement_by_class['label_categoria'] = agreement_by_class['categoria'].map(LABELS)

    fig = px.bar(
        agreement_by_class,
        x='categoria',
        y='agreement_medio_pct',
        error_y='agreement_desvio_pct',
        text='agreement_medio_pct',  # texto = valor numérico
        title='Agreement médio por categoria (com desvio padrão)',
        color='categoria',
        color_discrete_sequence=px.colors.qualitative.Set2,
    )

    fig.update_traces(
        texttemplate='%{text:.1f}%',  # uma casa decimal + %
        textposition='outside',
        hovertemplate=(
            '<b>%{customdata[0]}</b><br>'
            'Agreement médio: %{y:.1f}%<br>'
            'Desvio padrão: %{customdata[1]:.1f}%'
            '<extra></extra>'
        ),
        customdata=agreement_by_class[['label_categoria', 'agreement_desvio_pct']].values
    )

    y_max = agreement_by_class['agreement_medio_pct'].max()
    y_err = agreement_by_class['agreement_desvio_pct'].fillna(0).max()

    fig.update_layout(
        template='simple_white',
        showlegend=False,  # sem legenda
        yaxis_title='Agreement médio (%)',
        xaxis_title='Categoria',
        title_x=0.5,
        uniformtext_minsize=10,
        uniformtext_mode='hide',
        margin=dict(l=40, r=20, t=60, b=80),
        height=500,
        xaxis=dict(
            tickangle=-30,
            tickmode='array',
            tickvals=agreement_by_class['categoria'],
            ticktext=agreement_by_class['label_categoria'],
            categoryorder='array',
            categoryarray=agreement_by_class['categoria'].tolist()
        ),
        yaxis=dict(
            range=[0, min(100, y_max + y_err * 1.25)]
        )
    )

    fig.show()
else:
    print('Sem dados de agreement por categoria para exibir.')
```
