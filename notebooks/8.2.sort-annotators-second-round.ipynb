```python
import pandas as pd
```

```python
annotations_path = 'data/filtered_first_stage_annotations.csv'
annotations_df = pd.read_csv(annotations_path)
annotations_df
```

```python
import random
import csv

N_SENTENCES = annotations_df.shape[0]  # 16166
N_ANNOTATORS = 21
MIN_LABELS_PER_SENT = 2  

sentence_ids = annotations_df['frase_id'].tolist()
random.seed(42)
random.shuffle(sentence_ids)

# Offsets coprimos com 21 para variar as duplas (evita sempre as mesmas)
#    Todos abaixo são coprimos com 21 (21 = 3 * 7), então não múltiplos de 3 nem de 7.
OFFSETS = [1, 2, 4, 5, 8, 10]

assignments = []  

for idx, sent_id in enumerate(sentence_ids):
    a = idx % N_ANNOTATORS
    r = OFFSETS[(idx // N_ANNOTATORS) % len(OFFSETS)]
    b = (a + r) % N_ANNOTATORS

    assignments.append((sent_id, a))
    assignments.append((sent_id, b))
```

```python
from collections import Counter, defaultdict
load = Counter(ann for _, ann in assignments)
```

```python
load
```

```python
load_per_sentence = defaultdict(list)
for sent_id, ann_id in assignments:
    load_per_sentence[sent_id].append(ann_id)
load_per_sentence
```

```python
summary = {ann: load[ann] for ann in sorted(load)}
print("Cargas por anotador (id -> tarefas):")
print(summary)
total = sum(load.values())
print(f"Total de tarefas: {total} (esperado: 32332)")
print(f"Min={min(load.values())}, Max={max(load.values())}")
```

```python
assignments = [(sent_id, f"user{ann_id+1:02d}") for sent_id, ann_id in assignments]
assignments
```

```python
with open("data/assignments_second_round.csv", "w", newline="") as f:
    w = csv.writer(f)
    w.writerow(["frase_id", "annotator_id"])
    for row in assignments:
        w.writerow(row)
```
